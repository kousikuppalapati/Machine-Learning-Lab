```python
#import Libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#load the dataset
df=pd.read_csv('/content/cardata - cardata.csv')
```

```python
#count occurance of each Fuel_type in the dataset
df['Fuel_Type'].value_counts()
```

```python
#count occurance of each Seller_type in the dataset
df['Seller_Type'].value_counts()
```

```python
#count occurance of each Transmssion in the dataset
df['Transmission'].value_counts()
```

```python
#one hot encoding
df=pd.get_dummies(df,columns=['Fuel_Type','Seller_Type','Transmission'])
```

```python
df.columns
```

```python
#create feature matrix x b dropping car_name and selling-price columns
X=df.drop(['Car_Name','Selling_Price'],axis=1)
print("Feature matrix X created by dropping 'car_name' and 'selling_price' columns")
```

```python
#select target vector Y containing the selling_price column
Y=df['Selling_Price']
print("Target vector Y created by selecting 'selling_price' column")
```

```python
X.head()
```

```python
Y.head()
```

```python
#split the dataset into training and testing
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
print("Dataset split into training and testing sets")
```

```python
#create model
model=LinearRegression()
print("Linear regression model created")
```

```python
#train the model
model.fit(X_train,Y_train)
print("Model trained on training data")
```

```python
#Generate predictions for training and testing data
Y_train_pred=model.predict(X_train)
Y_test_pred=model.predict(X_test)
print("Predictions generated for training and testing data")
```

```python
#Evaluate the model performance on training data using metrics
train_mae=mean_absolute_error(Y_train,Y_train_pred)
train_mse=mean_squared_error(Y_train,Y_train_pred)
train_rmse=np.sqrt(train_mse)
train_r2=r2_score(Y_train,Y_train_pred)
print("Training metrics:")
print("Mean Absolute Error:",train_mae)
print("Mean Squared Error:",train_mse)
print("Root Mean Squared Error:",train_rmse)
print("R-squared:",train_r2)
```

```python
##Evaluate the model performance on testing data using metrics
test_mae=mean_absolute_error(Y_test,Y_test_pred)
test_mse=mean_squared_error(Y_test,Y_test_pred)
test_rmse=np.sqrt(test_mse)
test_r2=r2_score(Y_test,Y_test_pred)
print("Testing metrics:")
print("Mean Absolute Error:",test_mae)
print("Mean Squared Error:",test_mse)
print("Root Mean Squared Error:",test_rmse)
print("R-squared:",test_r2)
```

```python
#coefficient and Intercept
print("Slope:",model.coef_)
print("Intercept:",model.intercept_)
```
